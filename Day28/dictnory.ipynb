```python
d1={"A":"IT officer","B":"GST Inspector","C":"CBI"}
print(f"Original dictionary: {d1}")
all_values = d1.values()
print(f"All values in dictionary: {all_values}")
all_keys = d1.keys()
print(f"All keys in dictionary: {all_keys}")
```
